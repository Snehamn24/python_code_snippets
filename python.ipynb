# Write a program to accept a string from the user and count the number of vowels, consonants, digits and special characters.

```python
vow = ['a','e','i','o','u']
digits = [str(i) for i in range(0,10)]
sc=['@','#','$','%','^','&','*','(',')']

v=0
d=0
s=0
con=0

str1 = input("Enter the string : ").lower()
for c in str1:
  if c in vow:
    v+=1
  elif c in digits:
    d+=1
  elif c in sc:
    s+=1
  else:
    con+=1
print("Vowels : ",v)
print("Digits : ",d)
print("Special character : ",s)
print("Consonants : ",con)
```
